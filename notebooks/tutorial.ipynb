# CATIA Tutorial

This notebook runs a short end-to-end catastrophe risk analysis with CATIA.

```python
from catia.data_acquisition import fetch_all_data
from catia.risk_prediction import train_risk_model
from catia.financial_impact import run_multi_peril_analysis
from catia.mitigation import generate_mitigation_recommendations

region = "US_Gulf_Coast"
perils = ["hurricane", "flood"]

data = fetch_all_data(region, use_mock=True, perils=perils)
print(f"Climate records: {len(data['climate'])}, Events: {len(data['historical_events'])}")
```

```python
predictor = train_risk_model(
    data["climate"],
    data["socioeconomic"],
    data["historical_events"],
)
print("Risk model trained.")
```

```python
results = run_multi_peril_analysis(perils=perils)
agg = results["aggregate_metrics"]
print(f"Mean annual loss: ${agg['descriptive_stats']['mean']:,.0f}")
print(f"VaR (95%): ${agg['risk_metrics']['var']:,.0f}")
print(f"TVaR (95%): ${agg['risk_metrics']['tvar']:,.0f}")
```

```python
recs = generate_mitigation_recommendations(baseline_loss=agg["descriptive_stats"]["mean"])
print(f"Risk reduction: {recs['summary']['total_risk_reduction']:.2%}")
print("Priority order:", recs["priority_order"][:3])
```
